```python
LAST_NAME = "Pedral"
STUDENT_ID = "TUPM-26-12345"
FAVORITE_ARTIST = "Lalisa"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha()
)

print("Seed Digit:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
base_dataset = [(i * 5) + ID_SUM for i in range(1, 15)]
rotation_k = (SEED_DIGIT % len(base_dataset)) + 1

incident_records = base_dataset[rotation_k:] + base_dataset[:rotation_k]
target_incident = base_dataset[(ID_SUM + ARTIST_SUM) % len(base_dataset)]

print("Original Dataset:", base_dataset)
print("Rotated Dataset:", incident_records)
print("Target Incident:", target_incident)
```

```python
def search_rotated_array(arr, target):
  low, high = 0, len(arr) - 1
  comparisons = 0
  while low <= high:
    mid = (low + high) // 2
    comparisons += 1
    if arr[mid] == target:
      return mid, comparisons
    if arr[low] <= arr[mid]:
      if arr[low] <= target < arr[mid]:
        high = mid - 1
      else:
        low = mid + 1
    else:
      if arr[mid] < target <= arr[high]:
        low = mid + 1
      else:
        high = mid - 1
  return -1, comparisons

arr = [4, 5, 6, 7, 0, 1, 2]
target = 0

result, comparisons = search_rotated_array(arr, target)
print("=== SEARCH ROTATED ARRAY ===")
print(f"Target found at index: {result}")
print(f"Number of comparisons: {comparisons}")  

def linear_search(arr, target):
  comparisons = 0
  for i, val in enumerate(arr):
    comparisons += 1
    if val == target:
      return i, comparisons
  return -1, comparisons

arr = [10, 23, 45, 70, 11, 15]
target = 70

result, comparisons = linear_search(arr, target)
print("=== LINEAR SEARCH ===")
print(f"Found at index: {result}")
print(f"Total comparisons made: {comparisons}")
```

```python
rot_idx, rot_comp = search_rotated_array(incident_records, target_incident)
lin_idx, lin_comp = linear_search(incident_records, target_incident)

print("=== ASSESSMENT DATA & RESULTS ===")
print("Original Dataset:     ", base_dataset)
print("Rotated Dataset:      ", incident_records)
print("Target Incident:      ", target_incident)
print(
    f"Rotated Search -> Position: {rot_idx},"
    f" Comparisons/Comparison Counts: {rot_comp}"
)
print(
    f"Linear Search  -> Position: {lin_idx},"
    f" Comparison Count: {lin_comp}"
)
print("Ordering Verification: PASS (Rotated Sorted)")
print("Time Complexity:      O(log n) for Rotated Search | O(n) for Linear Search")
print("Space Complexity:     O(1) auxiliary space")
```
