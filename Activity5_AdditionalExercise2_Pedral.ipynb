```python
LAST_NAME = "Pedral"
STUDENT_ID = "TUPM-26-12345"
FAVORITE_ARTIST = "Lalisa"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha()
)

print("Seed Digit:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
dataset_A = sorted(
    [
        (SEED_DIGIT + 1) * 3,
        ID_SUM + 5,
        ID_SUM + 12,
        (ID_SUM + SEED_DIGIT) * 2,
    ]
)

dataset_B = sorted(
    [
        (ARTIST_LENGTH + NAME_LENGTH) * 2,
        (ARTIST_LENGTH * 3) + 4,
        (ARTIST_LENGTH + NAME_LENGTH) * 4,
    ]
)

print("Department A dataset:", dataset_A)
print("Department B dataset:", dataset_B)
```

```python
def merge_sorted_records(arr1, arr2):
  i = j = 0
  merged = []
  comparisons = 0

  while i < len(arr1) and j < len(arr2):
    comparisons += 1
    if arr1[i] <= arr2[j]:
      merged.append(arr1[i])
      i += 1
    else:
      merged.append(arr2[j])
      j += 1

  while i < len(arr1):
    merged.append(arr1[i])
    i += 1

  while j < len(arr2):
    merged.append(arr2[j])
    j += 1

  return merged, comparisons

arr1 = [1, 3, 5]
arr2 = [2, 4, 6]

merged_list, comparisons = merge_sorted_records(arr1, arr2)

print(f"Merged Array: {merged_list}")
print(f"Comparisons made: {comparisons}")
```

```python
consolidated_dataset, merge_comparisons = merge_sorted_records(
    dataset_A, dataset_B
)

is_properly_ordered = consolidated_dataset == sorted(consolidated_dataset)

print("=== ASSESSMENT DATA & RESULTS ===")
print("Department A dataset:              ", dataset_A)
print("Department B dataset:              ", dataset_B)
print("Consolidated dataset:              ", consolidated_dataset)
print(
    "Number of records before/after:    "
    f"{len(dataset_A) + len(dataset_B)} before ->"
    f" {len(consolidated_dataset)} after"
)
print("Comparison count:                  ", merge_comparisons)
print(
    "Ordering verification:             "
    f"{'PASS (Sorted)' if is_properly_ordered else 'FAIL'}"
)
print("Time complexity:                   O(n + m)")
print("Space complexity:                  O(n + m) auxiliary space")
```
