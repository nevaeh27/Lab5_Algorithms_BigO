```python
LAST_NAME = "Pedral"
STUDENT_ID = "TUPM-26-12345"
FAVORITE_ARTIST = "Lalisa"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha()
)

print("Seed Digit:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
base_orders = [
    (SEED_DIGIT * 7) % 100,
    (ID_SUM * 3) % 100,
    (NAME_LENGTH * 11) % 100,
    (ARTIST_LENGTH * 9) % 100,
    (ARTIST_SUM * 5) % 100,
]

conditions = {
    "Sorted (Best Case for some)": sorted(base_orders),
    "Reverse-ordered (Worst Case)": sorted(base_orders, reverse=True),
    "Random / Unsorted": base_orders.copy(),
}

print("Base Order Dataset:", base_orders)
for cond_name, dataset in conditions.items():
  print(f"{cond_name}: {dataset}")
```

```python
def bubble_sort(arr):
  a = arr.copy()
  comparisons = swaps = 0
  n = len(a)
  for i in range(n):
    swapped = False
    for j in range(0, n - i - 1):
      comparisons += 1
      if a[j] > a[j + 1]:
        a[j], a[j + 1] = a[j + 1], a[j]
        swaps += 1
        swapped = True
    if not swapped:
      break
  return a, comparisons, swaps

sample_arr = [64, 34, 25, 12, 22, 11, 90]

sorted_arr, comps, swaps = bubble_sort(sample_arr)
print("Bubble Sort Result:", sorted_arr)
print(f"Comparisons: {comps}, Swaps: {swaps}\n")  
```

```python
def selection_sort(arr):
  a = arr.copy()
  comparisons = swaps = 0
  n = len(a)
  for i in range(n):
    min_idx = i
    for j in range(i + 1, n):
      comparisons += 1
      if a[j] < a[min_idx]:
        min_idx = j
    if min_idx != i:
      a[i], a[min_idx] = a[min_idx], a[i]
      swaps += 1
  return a, comparisons, swaps

sample_arr = [64, 34, 25, 12, 22, 11, 90]

sorted_arr, comps, swaps = selection_sort(sample_arr)
print("Selection Sort Result:", sorted_arr)
print(f"Comparisons: {comps}, Swaps: {swaps}\n")
```

```python
def insertion_sort(arr):
  a = arr.copy()
  comparisons = shifts = 0
  for i in range(1, len(a)):
    key = a[i]
    j = i - 1
    while j >= 0:
      comparisons += 1
      if a[j] > key:
        a[j + 1] = a[j]
        shifts += 1
        j -= 1
      else:
        break
    a[j + 1] = key
  return a, comparisons, shifts

sample_arr = [64, 34, 25, 12, 22, 11, 90]

sorted_arr, comps, shifts = insertion_sort(sample_arr)
print("Insertion Sort Result:", sorted_arr)
print(f"Comparisons: {comps}, Shifts: {shifts}")
```

```python
import time

print("=== ASSESSMENT DATA & PERFORMANCE COMPARISON ===")
header = f"{'Condition':<30} | {'Algorithm':<15} | {'Comparisons':<12} | {'Movements':<12}"
print(header)
print("-" * 78)

for cond_name, dataset in conditions.items():
  # Bubble Sort
  _, b_comp, b_swap = bubble_sort(dataset)
  # Selection Sort
  _, s_comp, s_swap = selection_sort(dataset)
  # Insertion Sort
  _, i_comp, i_shift = insertion_sort(dataset)

  print(
      f"{cond_name:<30} | {'Bubble Sort':<15} | {b_comp:<12} | {b_swap:<12}"
  )
  print(
      f"{'':<30} | {'Selection Sort':<15} | {s_comp:<12} | {s_swap:<12}"
  )
  print(
      f"{'':<30} | {'Insertion Sort':<15} | {i_comp:<12} | {i_shift:<12}"
  )
  print("-" * 78)

print("\n--- Big O & Case Analysis Observations ---")
print("1. Bubble Sort Best-Case: O(n) [optimized with early exit]; Worst-Case: O(n^2)")
print("2. Selection Sort: Always performs O(n^2) comparisons regardless of initial order.")
print("3. Insertion Sort Best-Case: O(n) on already sorted inputs; Worst-Case: O(n^2) on reverse-ordered inputs.")
print("Time Complexity across algorithms: O(n^2) general quadratic growth.")
print("Space Complexity: O(1) auxiliary space (in-place sorting).")
```
